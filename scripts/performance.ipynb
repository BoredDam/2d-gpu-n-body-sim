```python
import csv
from subprocess import run

for b in range(10, 180000, 400):
    run(['./naive_parallel_for_testing.out', f'{b}', '10', 'galactic_chaos_180k.csv', '180ksweep_naive'])
```

```python
import matplotlib.pyplot as plt
import numpy as np

testname = "180ksweep_naive"
X = []
Y0 = []
Y1 = []
Y2 = []

with open(f"tests/{testname}/{testname}.csv", 'r') as csvfile:
    reader = csv.reader(csvfile)
    for d in reader:
        X.append(int(d[3]))
        Y0.append(float(d[0]))
        Y1.append(float(d[1]))
        Y2.append(float(d[2]))

plt.plot(X, Y0)
plt.plot(X, Y1)
plt.plot(X, Y2)
plt.grid(alpha=0.4)
plt.xlabel("numero di corpi")
plt.ylabel("tempo in ms")
plt.title("tempo di calcolo dei kernel nell'algoritmo naive (10 iterazioni)")
plt.legend(["calcolo posizioni", "calcolo velocità", "calcolo accelerazione"])
plt.savefig("./images/naive_kernel_ms.pdf")
plt.show()
```
